### 4. A program that prints the multiplication table of a number (from 1–12)

````python
def multiplier():
    num = int(input("Please input the number you will like to print the multiplication table"))
    mult = 1
    print("\nThis is the multiplication table of", num, ".")
    while mult <= 12:
            print(str(num), " X ",  str(mult), " = ", str(num * mult))
            mult = mult + 1

multiplier()
````

### 5. A program that calculates the factorial of a number using a loop

````python
def factorial():
    num = int(input("Please input the factorial of the number your will like to check"))
    i = num - 1
    while i > 0:
        num = num * i
        i -= 1
    return num

factorial()
````

### 6. A program that finds the largest number in a given list.

````python
def largest():
    lst = list(map(int, input("Welcome. Please input the list of numbers you will like to check using space as a separator").split()))
    max_num = lst[0]
    for i in lst:
        if max_num < i:
            max_num = i
    return max_num

largest()
````

### 7. Python program that counts how many vowels are in a given string.

````python
def vowel_counter():
    string = str(input('Welcome to the vowel counter program. Please input your string let\'s count the vowel in it')).lower()
    vowel = ['a', 'e', 'i', 'o', 'u']
    i = 0
    count = 0
    for i in string:
        if i in vowel:
            count += 1
    return count

vowel_counter()
````

### 9. A program that reverses a string without using slicing ([::-1])

````python
def reverse_str():
    string = input("Hello user, please input the string you will like to reverse")
    new_str = ''
    i = len(string) - 1
    while i > 0:
        new_str += string[i]
        i -= 1
    new_str += string[0]
    return new_str

reverse_str()
````

### 10. A program that generates the first 10 numbers of the Fibonacci sequence.

````python
def fibonacci():
    fibo_seq = [0, 1]
    i = 10
    for i in range(10):
        fibo_seq.append((fibo_seq[-1] + fibo_seq[-2]))
    return fibo_seq

fibonacci()
````

### 1. Student Grades System

````python
# 1. Student Grading System
def stud_grade_system():
    print("Welcome to the Student Grades System!\n")
    score = list(map(int, input("Please input the mark for the 5 subjects using space as a seperator").split()))
    average = sum(score) / len(score)
    if (average >= 80):
        print("The student grade is A.")
    elif (average >= 60 and average < 80):
        print("The student grade is B.")
    elif (average >= 40 and average < 80):
        print("The student grade is C.")
    else:
        print("The student grade is F.")

stud_grade_system()
````

### 2. Simple Calculator

````python
#2. Simple Calculator
def calculator():
    print("DannyRalph Calculator ©")
    option = int(input('To perform the following operations:\nPress 1 for addition.\nPress 2 for subtraction.\nPress 3 for multiplication.\nPress 4 for division.\n'))
    if (option == 1):
        num = list(map(int, input("Please input the numbers you will like to add using space as seperator only:\n").split()))
        add = 0
        for i in num:
            add += i
        print("The addition of", str(num[:]), "is ", add)
    elif (option == 2):
        num = list(map(int, input("Please input the numbers you will like to subtract using space as seperator only:\n").split()))
        i = len(num) - 1
        sub = num[0]
        while i > 0:
            sub -= num[i]
            i -= 1
        print("The subtraction of", str(num[:]), "is ", sub)
    elif (option == 3):
        num = list(map(int, input("Please input the numbers you will like to multiply using space as seperator only\n").split()))
        i = len(num) - 1
        mult = num[0]
        while i > 0:
            mult *= num[i]
            i -= 1
        print("The multiplication of", str(num[:]), "is ", mult)
    elif (option == 4):
        num = list(map(int, input("Please input the numbers you will like to divide using space as seperator only:\n").split()))
        i = len(num) - 1
        div = num[0]
        while i > 0:
            div /= num[i]
            i -= 1
        print("The division of", str(num[:]), "is ", div)
    else:
        print("Wrong input. Re-run the program and follow the instructions")

calculator()
````

### 6. What is the difference between local and global variables? Provide an example.

### Local Variables
Local Variables are variables stored or declared within a particular block of code, such as functions or loops. They can only be used within the block in which they are stored and the memory allocated to such variables are erased once block of code is declared existed. Local variables can hold the same name with other variables in other code blocks without causing any conflict because of scope limitation.
#### Example
```python
def local():
    var = "I am a local variable within the local() function code block."
    print(local())
local()
```
`I am a local variable within the local() function code block`

### Global Variables
Global variables are variables declared outside the code block or scope of any function in a particular program. They are usually stored at the top of a program or in a separate file. They can be accessed and used in any part of the program, including code blocks such as functions, loops, etc. Programmers use global variables to store values that need to be accessed in multiple parts of the program.
#### Example
```python
global_var = "I am a global variable and I can be used anywhere within this program"
def local():
    print(global_var)
local()
```

`I am a global variable and I can be used anywhere within this program`

### 8. Explain the difference between == and is operators in Python.

```==``` operator is used to check the equality of two values. Whereas the ```is``` operator is used to check if two or more variables are stored in the same memory location.
- `==` operator is called equlaity operator
- `is` operator is called identity operator
-
Examples are shown in the code cell below:

````python
a = [2, 4, 7]
b = [2, 4, 7]
c = a

# Scenario 1: Check for Equality
if a == b:
    print("a and b are equal to each other")
elif a == c:
    print("a and c are equal to each other")
elif b == c:
    print("b and c are equal to each other")
else:
    print("None of the variables are equal")

# Scenario 2: Check if the variables are stored in the same memory location
if a is b:
    print("a and b are stored in same memory location")
elif a is c:
    print("a and c are stored in same memory location")
elif b == c:
    print("b and c are stored in same memory location")
else:
    print("None of the variables are store in the same location")
````

### 9. Write a program that counts the number of words in a given text.

````python
def count_words(txt = ''):
    word_lst = txt.split()
    count = 0
    i = len(word_lst)
    while i > 0:
        count += 1
        i -= 1
    print(count)

count_words('Hi I am Raphael a Data Scientist')
````

### 12. Explain the difference between for and while loops. Give an example of each.

### For Loop
For loops are loops used in programming to iterate over a sequence such as a string, list, tuple, dictionary and other iterable objects in order to achieve a particular purpose or carry out specific operations. It is used to repeat a particular block of code a specified number of times.

### While Loop
A while loop is used to iterate over a sequence like for loop does. However, a while loop iterates over a sequence or a specific block until the specified condition is false.
The example of a `for loop` and `while loop` can be seen in the code cells below:

````python
# Print 1 to 10 using a for loop
for i in range(1, 11):
    print(i)
````

````python
i = 10
while i > 0:
    if i == 1:
        print(f"I will run {i} last times until the specified condition becomes false")
    else:
        print(f"I will run {i} times until the specified condition becomes false")
    i -= 1
````

### 13. What are Python exceptions? Write a program that handles division by zero.

Python exceptions are unexpected errors that occurs while running a python program. Python allows for exception handling in order to effectively handle errors, manage code, and prevent programs from crashing abruptly.

Below is a program that divides any two numbers and handles division by zero error

````python
def divider(num1, num2):
    if num2 > 0:
        print(num1 / num2)
    else:
        raise ValueError("A number can not be divided by zero")

# Try the divider function
def try_divider(num1, num2):
    try:
        print(divider(num1, num2))
    except ValueError as e:
        print(e)

try_divider(30, 3)
````

### 14. Write a program to check if a string is a palindrome.

````python
def palindrome_checker(string = ''):

    # Write a program that check the reverse of a string
    def reverse_str(string = ''):
        new_str = ''
        i = len(string) - 1
        while i > 0:
            new_str += string[i]
            i -= 1
        new_str += string[i]
        return new_str

    # Use the reverse_str function to check if the reversed string is equal to the actual string.
    reverse_str = reverse_str(string)
    if string == reverse_str:
        print("The string is a Palindrome")
    else:
        print("The string is not a Palindrome")

palindrome_checker('9009')
````
